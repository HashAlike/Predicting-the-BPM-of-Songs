```python
train_df["BeatsPerMinute"].hist(bins=50)

plt.xlabel("Beats Per Minute")
plt.ylabel("Frequency")
plt.title("Beats Per Minute Distribution")
plt.show()
```

```python
train_df.corr(numeric_only=True)["BeatsPerMinute"].sort_values()
```

```python
corr = train_df.drop(columns=["id"]).corr()

plt.figure(figsize=(10, 8))
plt.imshow(corr, aspect="auto")
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
features = test_df.columns.drop("id")

comparison = pd.DataFrame({
    "train_mean": train_df[features].mean(),
    "test_mean": test_df[features].mean(),
    "train_std": train_df[features].std(),
    "test_std": test_df[features].std()
})

comparison
```
