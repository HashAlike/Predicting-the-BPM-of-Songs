```python
corr = train_df.drop(columns=["id"]).corr()

plt.figure(figsize=(10, 8))
plt.imshow(corr, aspect="auto")
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
features = test_df.columns.drop("id")

comparison = pd.DataFrame({
    "train_mean": train_df[features].mean(),
    "test_mean": test_df[features].mean(),
    "train_std": train_df[features].std(),
    "test_std": test_df[features].std()
})

comparison
```

```python
target_corr = (
    train_df
    .corr(numeric_only=True)["BeatsPerMinute"]
    .drop("BeatsPerMinute")
    .sort_values(key=abs, ascending=False)
)

target_corr
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor

X = train_df.drop(columns=["BeatsPerMinute", "id"])
y = train_df["BeatsPerMinute"]

model = HistGradientBoostingRegressor(
    max_iter=200,
    learning_rate=0.1,
    max_leaf_nodes=31,
    random_state=42
)

model.fit(X, y)
```

```python
train_df.drop(columns=["id", "BeatsPerMinute"]).hist(
    figsize=(14, 10),
    bins=40
)

plt.tight_layout()
plt.show()
```

```python
features = [
    "RhythmScore",
    "Energy",
    "MoodScore",
    "AudioLoudness"
]

for feature in features:
    plt.figure(figsize=(7, 5))

    plt.scatter(
        train_df[feature],
        train_df["BeatsPerMinute"],
        alpha=0.05,
        s=5
    )

    plt.xlabel(feature)
    plt.ylabel("BeatsPerMinute")
    plt.title(f"{feature} vs BeatsPerMinute")

    plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.scatter(
    train_df["id"],
    train_df["BeatsPerMinute"],
    alpha=0.05,
    s=5
)

plt.xlabel("id")
plt.ylabel("BeatsPerMinute")
plt.title("ID vs BeatsPerMinute")

plt.show()
```

```python
features = [
    "RhythmScore",
    "AudioLoudness",
    "VocalContent",
    "AcousticQuality",
    "InstrumentalScore",
    "LivePerformanceLikelihood",
    "MoodScore",
    "TrackDurationMs",
    "Energy"
]

for feature in features:
    train_df["bin"] = pd.qcut(train_df[feature], 10, duplicates="drop")

    grouped = train_df.groupby("bin", observed=True)["BeatsPerMinute"].mean()

    print(f"\n{feature}")
    print(grouped)
```

```python
predictions = model.predict(X)

plt.figure(figsize=(8, 6))

plt.scatter(
    y,
    predictions,
    alpha=0.05,
    s=5
)

plt.xlabel("Actual BeatsPerMinute")
plt.ylabel("Predicted BeatsPerMinute")
plt.title("Actual vs Predicted BPM")

plt.show()
```
